### 1. First Python Code / print()
The `print()` function is used to output text or variables to the screen. 
It is usually the first command you learn in any programming language.


```python
print('Hello, World!')
print('Welcome to Python!')
```

### 2. Variables
Variables are used to store data in a program. 
You don't need to declare their type beforehand; Python figures it out automatically.


```python
name = 'Alice'
age = 20
print('Name:', name)
print('Age:', age)
```

### 3. Data Types
Python has several basic data types: strings (text), integers (whole numbers), floats (decimals), and booleans (True/False).
You can check the type of a variable using the `type()` function.


```python
greeting = 'Hello'      # String
count = 10              # Integer
price = 19.99           # Float
is_active = True        # Boolean

print(type(greeting))
print(type(count))
print(type(price))
print(type(is_active))
```

### 4. Basic Operators / Math
Python can be used as a calculator. It supports basic operators like addition (`+`), subtraction (`-`), multiplication (`*`), and division (`/`).


```python
x = 10
y = 3

print('Addition:', x + y)
print('Subtraction:', x - y)
print('Multiplication:', x * y)
print('Division:', x / y)
```

### 5. Lists
A list is an ordered, changeable collection of items. 
You can access elements by their index, which starts at 0.


```python
fruits = ['apple', 'banana', 'cherry',100]
print(fruits[0])  # Access first element

fruits.append('orange')  # Add a new element
print(fruits)

fruits.remove('apple')
print(fruits)
```

### 6. Tuples
Tuples are ordered collections similar to lists but cannot be changed after creation (immutable).
They are useful when you want to ensure the data remains constant.


```python
colors = ("red", "green", "blue")
print(colors[0])
print(len(colors))

# Unpacking a tuple
r, g, b = colors
print("Green is:", g)
```

### 7. Sets
A set is an unordered collection of unique elements. 
It is useful for removing duplicates and checking membership quickly.


```python
numbers = {1, 2, 2, 3, 4, 4, 5}
print(numbers)  # Duplicates are removed

numbers.add(6)
print(numbers)  # Duplicates are removed

```

### 8. Dictionaries
A dictionary stores data in key-value pairs. 
You can use a unique key to easily access or update its corresponding value.


```python
student = {'name': 'John', 'age': 21, 'major': 'Computer Science'}
print(student['name'])

student['age'] = 22  # Update value
print(student)
```

### 9. if / elif / else
Conditional statements allow you to execute different blocks of code depending on whether a condition is True or False.


```python
score = 85

if score >= 90:
    print('Grade: A')
elif score >= 80:
    print('Grade: B')
else:
    print('Grade: C')
```

### 10. for Loop
A `for` loop is used to iterate over a sequence (like a list, tuple, or string).
It runs a block of code once for each item in the sequence.


```python
for i in range(5):
    print(i)
```

```python
for i in range(0, 10, 2):
    print(i)
```

```python
animals = ['cat', 'dog', 'bird']

for animal in animals:
    print(animal)
```

```python
animals = ['cat', 'dog', 'bird']

for i in range(len(animals)):
    print(animals[i])
```

### 11. while Loop
A `while` loop continues to execute a block of code as long as its condition remains True.


```python
count = 0

while count < 3:
    print('Count is:', count)
    count += 1
```

### 12. Functions
Functions are reusable blocks of code that perform a specific task.
You can pass data into them (arguments) and they can return a result.


```python
def greet(name):
    return 'Hello, ' + name + '!'

message = greet('Rhydham & Pushkar')
print(message)
```
