# Data Visualization with Matplotlib & Seaborn

## 1. Introduction

Data visualization is the graphical representation of data. It helps us understand trends, patterns, and outliers in our data easily.

### Matplotlib
A foundational library in Python for creating basic, customizable plots. It provides detailed, low-level control over every element of a visualization.

### Seaborn
A library built on top of Matplotlib. It makes it much easier to create statistical visualizations with beautiful default styles.

**Simple Difference:** Use Matplotlib for basic or highly controlled plotting. Use Seaborn for quicker, easier statistical visualizations.

## 2. Dataset

In this notebook, we will **not** use large, complex real-world datasets. 
Instead, we will manually create tiny datasets using simple Python lists and dictionaries. This ensures that you can easily see exactly how the numbers turn into a graph!

````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set a clean grid style for Seaborn plots
sns.set_theme(style="whitegrid")
````

## 3. Matplotlib Basics


```python
plt.figure(figsize=(6, 4)) # Set the size of the plot
plt.plot(x, y, color='blue', marker='o', linestyle='--', label='My Line')
plt.xlabel("X-Axis Label") # Add an X-axis label
plt.ylabel("Y-Axis Label") # Add a Y-axis label
plt.title("My Plot Title") # Add a title
plt.legend() # Show the legend (uses the 'label' argument)
plt.grid(True) # Turn on the grid
plt.show() # Display the plot
```

### Important Parameters:
- `figsize`: The width and height of the plot.
- `color`: The color of the line, dots, or bars.
- `marker`: The shape of data points (e.g., `'o'` for circle).
- `linestyle`: Style of the line (e.g., `'--'` for dashed).
- `label`: Name of the data series for the legend.
- `grid`: Shows grid lines on the plot.
- `legend`: Displays the legend box.

```
Line Styles	Marker Styles
'-' (Solid) 	'o' (Circle)
':' (Dotted)	's' (Square)
'-.' (Dash-dot)	'^' (Triangle up)
'--' (Dashed)	'd' (Diamond)
None (No line)	'x' (Cross)
```

## 4. Important Plots

### Matplotlib: Line Plot

A line plot connects individual data points with a continuous line.
**When to use:** To show trends or changes over an ordered sequence, like time.

````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperatures = [22, 24, 21, 25, 26]

plt.figure(figsize=(6, 4))
plt.plot(days, temperatures, marker='o', color='red', linestyle='--')
plt.title("Temperature over the Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.grid(False)
plt.show()
````

### Matplotlib: Bar Chart

A bar chart uses rectangular bars to represent values.
**When to use:** To compare numerical quantities across different distinct categories.

````python

students = ["Alice", "Bob", "Charlie", "David"]
scores = [85, 92, 78, 88]

plt.figure(figsize=(6, 4))
# aplha is transparency level
plt.bar(students, scores, color='skyblue', alpha=0.8)
plt.title("Student Test Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.show()
````

### Matplotlib: Histogram

A histogram groups numerical data into "bins" and shows how many values fall into each bin.
**When to use:** To see the distribution or shape of a single numerical variable.

````python

ages = [22, 23, 22, 25, 25, 25, 28, 29, 29, 30, 32, 35, 36, 40]

plt.figure(figsize=(6, 4))
# 'bins' controls how many groups the data is divided into
plt.hist(ages, bins=5, color='purple', alpha=0.7)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
````

### Matplotlib: Scatter Plot

A scatter plot displays individual data points as dots.
**When to use:** To look for a relationship or pattern between two numerical variables.

````python

study_hours = [1, 2, 3, 4, 5, 6]
exam_scores = [50, 55, 65, 70, 85, 90]

plt.figure(figsize=(6, 4))
plt.scatter(study_hours, exam_scores, color='green', marker='x')
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()
````

### Matplotlib: Pie Chart

A pie chart shows parts of a whole as slices of a circle.
**When to use:** To show simple proportions when you have a small number of categories.

````python

languages = ["Python", "Java", "C++", "JavaScript"]
users = [40, 25, 15, 20]

plt.figure(figsize=(5, 5))
plt.pie(users, labels=languages, autopct="%1.1f%%", colors=["#ff9999", "#66b3ff", "#99ff99", "#ffcc99"])
plt.title("Programming Language Popularity")
plt.show()
````

### Seaborn: Line Plot

Seaborn creates line plots easily from Pandas DataFrames.
**When to use:** To show trends over time.

````python

df_line = pd.DataFrame({
    "Year": [2018, 2019, 2020, 2021, 2022],
    "Revenue": [100, 150, 130, 200, 250]
})

plt.figure(figsize=(6, 4))
sns.lineplot(data=df_line, x="Year", y="Revenue", marker="o")
plt.title("Revenue Over Time")
plt.show()
````

### Seaborn: Bar Plot

**When to use:** To compare numerical values across categories. Seaborn makes it very easy to color by a specific group using the `hue` parameter.

### hue → separates data into different groups using colors.

````python
df_bar = pd.DataFrame({
    "City": ["NY", "NY", "LA", "LA"],
    "Season": ["Summer", "Winter", "Summer", "Winter"],
    "Tourists": [500, 300, 450, 400]
})

plt.figure(figsize=(6, 4))
# 'hue' groups the data by Season and gives them different colors
sns.barplot(data=df_bar, x="City", y="Tourists",hue="Season")
plt.title("Tourists by City and Season")
plt.show()
````

### Seaborn: Histplot

**When to use:** To view the distribution of a single variable. Seaborn adds a smooth curve (KDE) over the histogram easily.


### kde=True → adds a smooth curve that shows where the data is concentrated.
### It helps you understand the overall shape of the data distribution beyond the individual histogram bars.

````python
df_hist = pd.DataFrame({
    "Weights": [55, 60, 62, 62, 65, 65, 65, 70, 72, 75, 80]
})

plt.figure(figsize=(6, 4))
sns.histplot(data=df_hist, x="Weights", bins=5, kde=True, color="teal")
plt.title("Distribution of Weights")
plt.show()
````

### Seaborn: Scatter Plot

**When to use:** To check relationships between two numerical variables. Again, `hue` makes coloring categories effortless.

````python
df_scatter = pd.DataFrame({
    "Height": [150, 160, 170, 165, 180, 155],
    "Weight": [50, 60, 75, 65, 85, 55],
    "Gender": ["F", "F", "M", "F", "M", "F"]
})

plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_scatter, x="Height", y="Weight", hue="Gender")
plt.title("Height vs Weight by Gender")
plt.show()
````

### Seaborn: Box Plot

A box plot summarizes data using its median, quartiles, and highlights outliers (dots outside the box).
**When to use:** To compare the spread and distribution of numerical data across different groups.

````python
df_box = pd.DataFrame({
    "Class": ["A", "A", "A", "A", "B", "B", "B", "B"],
    "Scores": [70, 75, 80, 72, 85, 90, 88, 50] # 50 is an outlier in Class B
})

plt.figure(figsize=(6, 4))
sns.boxplot(data=df_box, x="Class", y="Scores", palette="pastel")
plt.title("Score Distribution by Class")
plt.show()
````

### Seaborn: Count Plot

A count plot simply counts how many times each category appears in your dataset (like a histogram for text/categorical data).
**When to use:** To visualize the frequency of categories.

````python
df_count = pd.DataFrame({
    "Fruit": ["Apple", "Apple", "Banana", "Apple", "Banana", "Cherry", "Cherry", "Cherry", "Cherry"]
})

plt.figure(figsize=(6, 4))
sns.countplot(data=df_count, x="Fruit", palette="viridis")
plt.title("Fruit Inventory Count")
plt.show()
````

## 5. Correlation

**Correlation** measures how two numerical variables move together:
- **Positive Correlation:** As X goes up, Y goes up.
- **Negative Correlation:** As X goes up, Y goes down.
- **Weak/No Correlation:** X and Y do not have a clear relationship.
- **Range:** Values range from -1.0 (strong negative) to +1.0 (strong positive). 0 means no correlation.

*Important:* **Correlation does NOT mean causation.**

````python
# Tiny dataset showing positive, negative, and random relationships
df_corr = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Exam_Score": [50, 60, 70, 80, 90],   # Positive correlation with Study_Hours
    "Video_Games": [5, 4, 3, 2, 1],       # Negative correlation with Study_Hours
    "Shoe_Size": [9, 8, 10, 9, 8]         # No real correlation
})

# Calculate correlation matrix
corr_matrix = df_corr.corr(numeric_only=True)
corr_matrix
````

### Heatmap

A heatmap uses colors to represent numbers in a grid.
**When to use:** Perfect for visualizing correlation matrices. `annot=True` writes the exact correlation number inside each box.

````python
plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
````

## 6. Subplots

Subplots allow you to display multiple graphs in a single figure grid.

- `fig`: The overall window/figure.
- `axes`: The individual grid boxes where plots are drawn.
- `figsize`: The size of the entire figure.

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

# Create a 2x2 grid (4 plots total)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Top-Left (Row 0, Col 0)
axes[0, 0].plot(x, y, color='blue')
axes[0, 0].set_title("Line Plot")

# Top-Right (Row 0, Col 1)
axes[0, 1].bar(x, y, color='orange')
axes[0, 1].set_title("Bar Chart")

# Bottom-Left (Row 1, Col 0)
axes[1, 0].scatter(x, y, color='green')
axes[1, 0].set_title("Scatter Plot")

# Bottom-Right (Row 1, Col 1)
axes[1, 1].hist(y, bins=3, color='purple')
axes[1, 1].set_title("Histogram")

plt.tight_layout() # Prevents overlapping
plt.show()
````

## 7. Basic Plot Selection

When should you use which plot?

| Plot | Best Used For |
|---|---|
| **Line** | Trends or changes over time |
| **Bar** | Comparing values across categories |
| **Histogram** | Viewing the distribution of numerical data |
| **Scatter** | Checking the relationship between two numerical variables |
| **Box** | Seeing data spread and finding outliers |
| **Count** | Finding the frequency of categories |
| **Pie** | Showing simple proportions / parts of a whole |
| **Heatmap** | Viewing correlation or matrix values |

### Useful Parameters
- `figsize`: Size of the figure (width, height).
- `color`: Color of the visual element.
- `marker`: Shape of the data point (e.g., `'o'`, `'x'`).
- `linestyle`: Line pattern (e.g., `'--'`).
- `label`: Name of the series for the legend.

- `annot`: Display values inside heatmap cells.
- `cmap`: Color scheme for heatmaps (e.g., `'coolwarm'`).
